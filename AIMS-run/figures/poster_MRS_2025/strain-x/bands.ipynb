```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
from turtle import title
from plotly.subplots import make_subplots
import plotly.tools as tls
from aimstools import BandStructure
from aimstools.bandstructures import RegularBandStructure as RBS
from aimstools.density_of_states import SpeciesProjectedDOS
from aimstools import DensityOfStates as DOS
from matplotlib import gridspec
import matplotlib.pyplot as plt
from inspect import signature
```

```python
bs_0 = RBS("/mnt/c/Users/marce/Documents/BPN/AIMS-run/strain/Strain-x/0.0%/input_files_HSE06_tight/input_files/")
dos_0 = DOS("/mnt/c/Users/marce/Documents/BPN/AIMS-run/strain/Strain-x/0.0%/input_files_HSE06_tight/input_files/")
```

```python
fig = plt.figure(constrained_layout=False, figsize=(21, 18))
spec = gridspec.GridSpec(ncols=2, nrows=1, figure=fig, width_ratios=[3.25, 0.75], wspace=0)

plt.rcParams.update({
    "font.size": 50,           # base font size
    "axes.labelsize": 35,      # x/y labels
    "xtick.labelsize": 25,     # tick labels
    "ytick.labelsize": 25,
    "legend.fontsize": 25
})

ax1 = fig.add_subplot(spec[0])
ax1 = bs_0.plot(axes=ax1, 
              color="blue", 
              main=True, 
              window = (-8 , 8) , 
              fermi_level_color =  "red",
              fermi_level_linewidth =  2.5,
              grid_linewidth = 3, 
              grid_linestyle = '-', 
              linewidth = 2.5)

ax1.set_xlabel("k-points")
ax1.tick_params(labelsize=25)

# The options to plot DOS are discussed in another tutorial. Here, we just plot all species.
ax2 = fig.add_subplot(spec[1])
ax2 = dos_0.total_dos.plot(axes=ax2, 
                         window = (-6 , 6),
                         fermi_level_color =  "red",
                         fermi_level_linewidth =  2.5,
                         broadening = 0.02)

ax2.set_yticks([])
ax2.set_xticks([1 , 2 , 3 , 4])
ax2.set_ylabel("")
ax2.set_xlabel("DOS")
#fig.suptitle("Band structure + DOS")
plt.savefig("/mnt/c/Users/marce/Documents/BPN/AIMS-run/figures/bands_0%.png" , dpi=200)
plt.show()
```
